```python
# Removing the suspected faued and canceled order status
df = df[
    (df['Order Status'] != 'SUSPECTED_FRAUD') &
    (df['Order Status'] != 'CANCELED')
]
```

```python
df.columns
```

```python
df=df.drop(columns=['Product Description'])
```

```python
df=df.drop(columns=['Product Image'])
```

```python
df=df.drop(columns=["Customer Email", "Customer Password"])
```

```python
df['Product Status'].values
# DROPING THE  COLUMNS AS ALL THE VALUES ARE 0 THUS VARIANCE IS 0 AND CAN'T GET ANY INSIGHTS FROM IT 
```

```python
# REMOVING DUPLICATE COLUMNS 
#Product Price == Order Item Product Price
df=df.drop(columns=["Product Price", "Product Card Id"])
```

```python
df=df.drop(columns=['Customer Fname','Customer Lname','Order Profit Per Order'])
```

# HANDLING MISSING VALUES 

```python
df["Order_Zipcode_missing"] = df["Order Zipcode"].isna()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(
    x="Order_Zipcode_missing",
    y="Sales",
    data=df
)

plt.xlabel("Order Zipcode Missing")
plt.ylabel("Sales")
plt.show()
```

```python
numeric_cols = [
    "Sales",
    "Order Item Quantity",
    "Order Item Product Price",
    "Order Item Discount Rate",
    "Order Item Total"
]

for col in numeric_cols:
    plt.figure(figsize=(6, 4))

    sns.boxplot(
        x="Order_Zipcode_missing",
        y=col,
        data=df
    )

    plt.title(f"{col} vs Order Zipcode Missingness")
    plt.xlabel("Order Zipcode Missing")
    plt.ylabel(col)

    plt.show()
```

# Conclusion 

```python
#Missing values are completly at random MCAR so applying CCA
df=df.drop(columns=["Order Zipcode"])
```

```python
mode_zip = df["Customer Zipcode"].mode()[0]
df["Customer Zipcode"] = df["Customer Zipcode"].fillna(mode_zip)
```

```python
df["Order_Time"].isnull().sum()
```

```python
df["Order_Time"] = df["Order_Time"].fillna(method="ffill")
```

# EDA

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size for readability
plt.figure(figsize=(10, 8))

# Generate the optimized correlation heatmap
sns.heatmap(
    df.corr(numeric_only=True),
    annot=False,                 
    fmt=".2f",                   
    cmap="coolwarm",        
    vmin=-1,                    
    vmax=1,                    
    square=True,                
    linewidths=0.5            
)

# Display the plot
plt.title("Correlation Matrix Heatmap", fontsize=14, pad=15)
plt.show()
```
